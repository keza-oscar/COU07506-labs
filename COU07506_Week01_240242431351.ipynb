## Task 1.1 — My first notebook: elementary sequences

```python
n = np.arange(-5, 11)
delta = (n == 0).astype(float)      # unit impulse
step = (n >= 0).astype(float)       # unit step
ramp = n * step                     # unit ramp
expo = 0.8**n * step                # real exponential

fig, axes = plt.subplots(1, 4, figsize=(13, 2.6))
for ax, x, t in zip(axes, [delta, step, ramp, expo],
                   ['delta(n)', 'u(n)', 'r(n)', '(0.8)^n u(n)']):
    ax.stem(n, x, basefmt='k-')
    ax.set_title(t); ax.set_xlabel('n (samples)'); ax.set_ylabel('amplitude')
plt.tight_layout(); plt.show()
```

## Task 1.2 — Change the parameters

(a) Plot $a^n u(n)$ for $a=0.5,-0.8,1.1$.  
(b) Plot $\cos(2\pi n/8)$ and $\cos(2\pi n/7.5)$ for $0\le n\le 40$.

**Answer:** $\cos(2\pi n/8)$ repeats exactly (period $N=8$, an integer). $\cos(2\pi n/7.5)$ needs $N=7.5$ samples, which is not an integer, so it is **not** periodic in discrete time (explained in Week 3).

```python
n = np.arange(0, 41)
fig, axes = plt.subplots(1, 3, figsize=(13, 3), sharex=True)
for ax, a in zip(axes, [0.5, -0.8, 1.1]):
    ax.stem(n, a**n, basefmt=' ')
    ax.set_title(f'a = {a}'); ax.set_xlabel('n (samples)'); ax.set_ylabel('a^n')
plt.tight_layout(); plt.show()

plt.figure(figsize=(9, 3.5))
plt.stem(n, np.cos(2*np.pi*n/8), basefmt=' ', label='cos(2 pi n / 8), N = 8')
plt.stem(n, np.cos(2*np.pi*n/7.5), linefmt='C1--', markerfmt='C1o', basefmt=' ',
         label='cos(2 pi n / 7.5), N = 7.5')
plt.title('Periodic vs non-periodic discrete-time cosine'); plt.xlabel('n (samples)')
plt.ylabel('amplitude'); plt.legend(); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
```

## Task 1.3 — Working with audio

Put your phone recording in this folder as `my_voice.wav` (3–5 s). The fallback signal is used only if the file is missing.

```python
import os
fname = 'my_voice.wav'
if os.path.exists(fname):
    fs, x = wavfile.read(fname)
    if x.ndim > 1:
        x = x.mean(axis=1)
    x = x / np.max(np.abs(x))
else:
    fs = 16000
    t = np.arange(0, 2, 1/fs)
    x = 0.5*np.sin(2*np.pi*440*t)*np.exp(-t)
    print('FALLBACK SIGNAL - replace with my_voice.wav')
t = np.arange(len(x)) / fs
print(f'fs = {fs} Hz, duration = {len(x)/fs:.2f} s, samples = {len(x)}')
plt.plot(t, x); plt.xlabel('time (s)'); plt.ylabel('amplitude (normalised)')
plt.title('My recording'); plt.grid(alpha=0.3); plt.tight_layout(); plt.show()
# To listen: Audio(x, rate=fs)
```
